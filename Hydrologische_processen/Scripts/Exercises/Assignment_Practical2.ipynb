<strong><u>While loop</u></strong>

Een <b>while-loop</b> voert een blok code herhaaldelijk uit zolang een bepaalde voorwaarde waar (`True`) is.

Algemene syntax:

```python
while voorwaarde:
    code
```

Werking:

1. De voorwaarde wordt gecontroleerd.
2. Als de voorwaarde `True` is, wordt de code uitgevoerd.
3. Na de uitvoering wordt de voorwaarde opnieuw gecontroleerd.
4. Dit herhaalt zich totdat de voorwaarde `False` wordt.

Voorbeeld:

```python
i = 1
while i <= 5:
    print(i)
    i += 1 # hetzelfde als i = i + 1
```
Uitvoer:

```python
1
2
3
4
5
```
-> Zolang <code>i</code> kleiner of gelijk is aan 5, zal er telkens 1 worden opgeteld bij <code>i</code>.

-> Wanneer <code>i = 6 </code>, <code>(i <= 5) = FALSE </code> dus de while-loop wordt niet meer uitgevoerd.

<strong><u>If-statement</u></strong>

Een <b>if-statement</b> voert een blok code éénmalig uit als een bepaalde voorwaarde waar (`True`) is.

Algemene syntax:

```python
if voorwaarde_1:
    code_1
elif voorwaarde_2:
    code_2
else: 
    code_3
```

Werking:

1. De voorwaarde wordt gecontroleerd.
2. Als voorwaarde_1 `True` is, wordt de code_1 uitgevoerd en de rest van de if-codeblok wordt niet uitvoerd.
3. Als voorwaarde_1 `False`is, wordt gekeken naar voorwaarde_2. 
4. Als voorwaarde_2 `True` is, wordt de code_2 uitgevoerd en de rest van de if-codeblok wordt niet uitvoerd.
5. Als voorwaarde_2 `False`is, wordt code_3 uitgevoerd. 

`elif` en `else` zijn optioneel, je kan ook een if-statement maken met enkel if en voorwaarde_1 en code_1 of één van beide toevoegen.

````python
def ModeleerAfvoer(
    date_ini:pd.Timestamp = pd.to_datetime("2024-07-09 12:00:00.000"),
    date_end:pd.Timestamp = pd.to_datetime("2024-07-14 16:00:00.000"),
    df:pd.DataFrame = df.copy(),
    df_UH:pd.DataFrame = df_UH.copy(),
    phi: float=phi,
    Q_base: float=Q_base,
    Oppervlakte: float=CATCHMENT_AREA_DEMER) -> pd.DataFrame:

    # Selecteer de periode
    df_q5 = ...

    phi_m3 = ...

    # Selecteer alleen de data van momenten met effectieve regen

    effective_rainevent = ...
    effective_rainevent['Peff (m3/h)'] = ...

    def simoutput (effective_rainevent, i):
        # Extraheert het i-de uur neerslag uit het dataframe effective_rainevent 
        effectieve_neerslag = effective_rainevent['Peff (m3/h)'].iloc[i]
        # Berekent de directe afvoer van het i-de uur neerslag o.b.v. het eenheidshydrogram
        afvoer = effectieve_neerslag*df_UH['UH']

        # Extraheert het uur van het i-de uur neerslag
        starttijd = effective_rainevent["Datum_tijd"].iloc[i]
        # Maakt een tijdsreeks startende van de starttijd o.b.v. het eenheidshydrogram
        tijd = starttijd + pd.to_timedelta(df_UH["time_after_rain"], unit="h")

        # Maakt een dataframe met de tijd en de gesimuleerde directe afvoer
        df_sim = pd.DataFrame({"Datum_tijd" : tijd, 
                               "Directe Afvoer Simulatie (m3/h)" : afvoer})
        return df_sim

    # Simuleert de directe afvoer van ieder uur van effectieve neerslag en voegt de output van iedere uur effectieve neerslag samen in 1 dataframe
    sim = pd.concat([simoutput(effective_rainevent, i) for i in range(effective_rainevent.shape[0])])

    # Indien er maar één uur effectieve neerslag is, is het dataframe van hiervoor de output.
    # Indien er meer dan één uur effectieve neerslag is, moet deze gegroepeerd worden per uur en gesommeerd over de verschillende uren effectieve neerslag
    if effective_rainevent.shape[0] > 0:
        sim_df = sim.groupby(['Datum_tijd'])['Directe Afvoer Simulatie (m3/h)'].sum().reset_index()
    else:
        sim_df = sim

    # Berekent de afvoer door de toevoeging van de basisafvoer
    sim_df['Afvoer Simulatie (m3/h)'] = sim_df['Directe Afvoer Simulatie (m3/h)'] + Q_base

    # Maakt een dataframe met de tijd, de simulatie van de afvoer en de geobserveerde afvoer en neerslag
    data = pd.merge(sim_df, df[['Datum_tijd', 'Q (m3/h)', 'P (mm/h)']], on=['Datum_tijd'], how="inner")
    data = data.rename(columns={'Q (m3/h)' : 'Afvoer Observatie (m3/h)'})
    
    print(data.head(5))
    return(data)
````
