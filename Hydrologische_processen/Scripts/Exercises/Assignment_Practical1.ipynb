**<u>Korte samenvatting van de betreffende theorie</u>**

De water cyclys weergeven met de  <b>waterbalans</b> vergelijking van een stroomgebied:

$Neerslag (P) = Evapotranspiratie (E) + Afvoer (Q) + Verandering bergin (\Delta S)$

<b>Klimatologie</b>: meerjarig, langdurig gemiddelde

<b>Anomalie</b>: verschil met langdurig gemiddelde


<b>Aflijnen van een stroomgebied aan de hand van een digitaal hoogte model</b>

Een stroomgebied kan worden bepaald met behulp van een Digital Elevation Model (DEM). 
Door opeenvolgende ruimtelijke analyses wordt berekend hoe water over het landschap 
stroomt en waar rivieren ontstaan.

Het proces bestaat uit vier belangrijke stappen:

1. Het DEM beschrijft de terreinhoogte.
2. De drainagerichting bepaalt waar water naartoe stroomt.
3. De cumulatieve drainage toont waar water zich verzamelt.
4. Een threshold selecteert de belangrijkste afvoerbanen en vormt het riviernetwerk.

**DEM → Flow Direction → Flow Accumulation → River Network → Stroomgebied**

---

1. DEM (Digital Elevation Model)

Een DEM bevat de hoogtewaarden van het terrein. 
Elke rastercel heeft een bepaalde elevatie (bijvoorbeeld in meter).

Een voorbeeld van een vereenvoudigd DEM:

|   | A | B | C | D | E |
|---|---|---|---|---|---|
| 1 | 120 | 115 | 110 | 105 | 100 |
| 2 | 125 | 118 | 112 | 102 | 95 |
| 3 | 130 | 122 | 108 | 98 | 90 |
| 4 | 135 | 126 | 115 | 100 | 85 |
| 5 | 140 | 132 | 120 | 105 | 80 |

De hogere gebieden bevinden zich linksboven, terwijl de lagere gebieden 
zich rechtsonder bevinden. Water zal dus naar de lagere zones stromen.

---

2. Drainagerichting (Flow Direction)

In deze stap wordt voor elke rastercel bepaald naar welke buurcel het water stroomt.

Het water stroomt steeds naar de cel met de grootste daling in hoogte.

Voorbeeld:

|   | A | B | C | D | E |
|---|---|---|---|---|---|
| 1 | ↘ | → | ↘ | ↓ | ↓ |
| 2 | → | ↘ | ↘ | ↓ | ↓ |
| 3 | → | → | ↘ | ↘ | ↓ |
| 4 | → | ↘ | → | ↘ | ↓ |
| 5 | → | → | → | → | O |

Legenda:

- `→ ↓ ↘` = richting waarin water afstroomt
- `O` = uitstroompunt (outlet)

Deze kaart beschrijft dus het potentiële afvoerpad van regenwater.

---

3. Cumulatieve drainage (Flow Accumulation)

De flow accumulation berekent hoeveel bovenliggende rastercellen bijdragen 
aan de afvoer naar een bepaalde cel.

Een hoge waarde betekent dat veel water samenkomt en dat er mogelijk een rivier 
aanwezig is.

Voorbeeld:

|   | A | B | C | D | E |
|---|---|---|---|---|---|
| 1 | 1 | 1 | 1 | 1 | 1 |
| 2 | 1 | 2 | 2 | 3 | 2 |
| 3 | 1 | 2 | 5 | 6 | 4 |
| 4 | 1 | 2 | 3 | 12 | 5 |
| 5 | 1 | 2 | 3 | 4 | 25 |

De hoogste waarden bevinden zich meestal langs de hoofdrivieren.

---

4. Threshold toepassen voor rivieren

Om rivieren te identificeren wordt een drempelwaarde (**threshold**) gekozen.

Bijvoorbeeld:


Alle cellen met een flow accumulation groter of gelijk aan 5 worden aangeduid 
als rivier.

Resultaat:

|   | A | B | C | D | E |
|---|---|---|---|---|---|
| 1 | . | . | . | . | . |
| 2 | . | . | . | . | . |
| 3 | . | . | R | R | . |
| 4 | . | . | . | R | R |
| 5 | . | . | . | . | R |

Legenda:

- `R` = riviercel
- `.` = geen rivier

<b>Bereken de maandelijkse waterbalans</b> aan de hand van de continuïteitsvergelijking en <b>maak twee grafieken</b> van de berging (S) [m³] in de tijd en van de verandering in berging (∆S=P-ET-Q) in de tijd. De continuïteitsvergelijking voor discrete tijdsstappen wordt gegeven door: 
</div>

$S_i = S_{i-1} + P - ET - Q$

*Hier geven we een voorbeeld van hoe je een functie in Python opbouwd, dit zorgt ervoor dat je makkelijker berekeningen meermaals kan toepassen*

````python
def volume_balk(l: float = 1, b: float = 1, h: float = 1) -> float:
    """
    Bereken het volume van een balk gegeven de lengte, breedte en hoogte.
    l = lengte, b = breedte, h = hoogte
    Indien je geen waarden meegeeft, wordt er standaard 1 m³ berekend a.d.h.v. de default waarden meegegeven hier 1.
    Hier wordt ook de tip gegeven om floats te gebruiken, zodat je ook decimalen kan meegeven.
    -> float wijst erop dat de functie een float teruggeeft.
    """
    volume = l * b * h
    return volume
````

````python
balk = volume_balk(10, 5, 2)
print(f"Volume of the balk: {balk} m³")
balk = volume_balk(10, 5.1)
print(f"Volume of the balk: {balk} m³")
balk = volume_balk()
print("Volume of the balk:", balk, "m³") # Hier een andere manier van printen zonder f-string
````

<b><u>Integer vs Float</u></b> 

Het zijn verschillende klassen

<ul>
<li><strong>Interger (int)</strong> = gehele getallen (zonder decimalen)</li>
<li><strong>Float (float)</strong> = kommagetallen (met decimalen)</li>
<ul>
    <li><strong>Float64</strong> = 64 bits, is de standaard met ~15-16 significante cijfers/li>
    <li><strong>Float32</strong> = 32 bits met ~7 significante cijfers (lagere accuraatheid)</li>

</ul>
</ul>

````python
def volume_balk_int(l: int = 1, b: int = 1, h: int = 1) -> int:
    """
    Bereken het volume van een balk gegeven de lengte, breedte en hoogte.
    l = lengte, b = breedte, h = hoogte
    Indien je geen waarden meegeeft, wordt er standaard 1 m³ berekend a.d.h.v. de default waarden meegegeven hier 1.
    Hier wordt ook de tip gegeven om integers te gebruiken, zodat je ook decimalen kan meegeven.
    -> int wijst erop dat de functie een int teruggeeft.
    """
    if not all(isinstance(x, int) for x in (l, b, h)): 
        # Hier wordt gecontroleerd of l, b en h integers zijn. Als dat niet zo is, wordt er een TypeError opgegooid.
        raise TypeError("l, b en h moeten integers zijn")
    volume = l * b * h
    return volume
````

````python
balk = volume_balk_int(10, 5, 2)
print(f"Volume of the balk: {balk} m³")
volume_balk_int(10.1, 5.2, 2.3)
print(f"Volume of the balk: {balk} m³")
````

<strong><u>Functie: groupby()</u></strong>

df is in het voorbeeld een dataframe
```python
df.groupby(group_columns)[value_columns].aggregation().rename(new_name).reset_index()
```
De verschillende argumenten van de functie:
<ol>
<li><b>group_columns</b>: In welke groepen wil ik de data verdelen?</li>
<li><b>value_columns</b>: Van welke kolommen wil ik berekeningen maken?</li>
<li><b>aggregation</b>: Welke berekening wil ik voor iedere groep maken? </li>
<ul>
    <li><b>min</b>: minimumwaarde binnen de groep</li>
    <li><b>max</b>: maximumwaarde binnen de groep</li>
    <li><b>mean</b>: gemiddelde waarde binnen de groep</li>
    <li><b>sum</b>: som van alle waarden binnen de groep</li>
    <li><b>count</b>: aantal niet-lege (niet-NaN) waarden in een kolom binnen de groep</li>
    <li><b>size</b>: totaal aantal rijen binnen de groep, inclusief rijen met ontbrekende waarden (NaN)</li>
    <li><b>...</b></li></ul>
    Wanneer je de bijvoorbeeld de som van een kolom van een dataframe wil, kan dat op dezelfde manier met: <code>df["kolom_naam"].sum()</code>
<li><b>rename</b>: (Optioneel) Hernoemt de naam van de uitvoer. De werking hangt af van het type object dat wordt teruggegeven:
    <ul>
        <li>Bij een <b>Series</b> (één geselecteerde kolom) geeft <code>.rename("nieuwe_naam")</code> een nieuwe naam aan de Series.</li>
        <li>Bij een <b>DataFrame</b> (één of meerdere geselecteerde kolommen) moet een kolomnaam worden hernoemd met <code>.rename(columns={"oude_naam": "nieuwe_naam"})</code> of <code>.rename(columns={"oude_naam_1": "nieuwe_naam_1", "oude_naam_2": "nieuwe_naam_2"})</code>.</li>
    </ul>
</li>
<li><b>reset_index</b>: Zet de group_columns terug om in normale kolommen zodat het een normaal dataframe is.</li>
<b>Let op</b>: Wanneer je groepeert op een kolom en hierop ook je berekeningen doet (zoals bij .count() en .size()), zal het resultaat van aggregation dezelfde naam krijgen. In deze gevallen moet je de nieuwe kolom een nieuwe naam geven anders kan je reset_index() niet toepassen want je hebt 2 kolommen met dezelfde naam.
</ol>
